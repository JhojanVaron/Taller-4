**1.** Entrenas un modelo y obtienes un 99% de exactitud sobre los datos de entrenamiento, pero solo un 75% sobre los datos de prueba. ¿Qué problema indica este resultado y por qué?

**RESPUESTA**

Este resultado indica claramente que el modelo sufre de overfitting o sobreajuste. La enorme diferencia entre el 99% de exactitud en entrenamiento y solo 75% en prueba demuestra que el modelo está memorizando los datos de entrenamiento en lugar de aprender patrones generalizables. Básicamente, el modelo se ajustó demasiado a las particularidades específicas del conjunto de entrenamiento. Para solucionar esto, se deberían aplicar técnicas de regularización como L1 o L2, usar dropout, implementar early stopping, o incluso reducir la complejidad del modelo y conseguir más datos para entrenar.

**2.** Si el error de tu modelo es muy alto tanto en el conjunto de entrenamiento como en el de validación, ¿cuál es el problema más probable? ¿Creerías que añadir más datos de entrenamiento solucionaría el problema?

**RESPUESTA**

Este resultado indica que el modelo sufre de underfitting o subajuste. Cuando el error es alto tanto en entrenamiento como en validación, significa que el modelo no está capturando adecuadamente los patrones de los datos. Es demasiado simple para la complejidad del problema. Respecto a añadir más datos, esto probablemente no solucionaría el problema porque la limitación no está en la cantidad de ejemplos sino en la capacidad del modelo para aprender. Las soluciones más efectivas serían aumentar la complejidad del modelo agregando más capas o neuronas, entrenar por más épocas, reducir la regularización si es excesiva, o hacer mejor feature engineering para capturar relaciones más complejas.

**3.** En un problema para predecir fallas en una máquina, tienes 100 variables provenientes de sensores, pero sospechas que solo unas pocas son realmente importantes. ¿Usarías Ridge o Lasso? Justifica tu respuesta.

**RESPUESTA**

En este caso, se debería usar Lasso en lugar de Ridge. La razón principal es que Lasso realiza selección automática de variables al reducir los coeficientes de las características irrelevantes exactamente a cero, eliminándolas del modelo. Esto es fundamental cuando se tienen 100 variables de sensores pero se sospecha que solo algunas son importantes. Ridge, por otro lado, nunca elimina variables completamente, solo reduce sus coeficientes hacia valores pequeños, manteniendo las 100 variables en el modelo. Con Lasso se obtendría un modelo más simple e interpretable, identificando quizás solo 10-15 sensores realmente predictivos de las fallas. Esto no solo mejora la interpretabilidad sino que también facilita el mantenimiento futuro al enfocarse únicamente en los sensores más relevantes para predecir problemas en la máquina.

**4.** Si entrenas un modelo Lasso y aumentas gradualmente el valor del hiperparámetro de penalización (λ), ¿qué efecto esperarías observar en los coeficientes del modelo?

**RESPUESTA**

Al aumentar gradualmente el valor de lambda en Lasso, los coeficientes del modelo se reducirían progresivamente hacia cero. Con lambda bajo, los coeficientes serían similares a una regresión normal. A medida que lambda aumenta, las variables menos importantes empezarían a llegar exactamente a cero y serían eliminadas del modelo, volviéndolo más simple y disperso. Las variables más relevantes resistirían valores más altos de lambda antes de desaparecer. Con lambda muy alto, casi todos los coeficientes llegarían a cero, resultando en un modelo que prácticamente no hace predicciones útiles. Este comportamiento permite identificar qué variables son más importantes según el orden en que son eliminadas al incrementar la penalización.

**5.** Al ejecutar el código de regularización 3D, ¿qué sucede con los coeficientes del modelo a medida que aumenta el valor de λ? ¿Qué interpretación le das a la forma diferente en que Ridge y Lasso aplican sus penalizaciones?

**RESPUESTA**

Al aumentar lambda, en Ridge los coeficientes se reducen gradualmente hacia cero de forma suave pero nunca llegan exactamente a cero, mientras que en Lasso los coeficientes caen abruptamente a cero uno por uno. La interpretación geométrica explica estas diferencias: Ridge usa una penalización L2 que forma una esfera en el espacio de parámetros, por lo que es difícil que los coeficientes toquen exactamente cero. Lasso usa penalización L1 que forma un diamante con esquinas en los ejes, haciendo que los puntos óptimos caigan frecuentemente en estas esquinas donde los coeficientes son exactamente cero, realizando así selección automática de variables.

**6.** Quieres optimizar un modelo Ridge y pruebas manualmente alpha=10, obteniendo un buen resultado. ¿Por qué sigue siendo metodológicamente superior usar GridSearchCV en lugar de quedarte con ese valor?

**RESPUESTA**

Usar GridSearchCV es metodológicamente superior porque evita el sesgo de selección al evaluar el hiperparámetro mediante validación cruzada sobre múltiples particiones de datos, no solo sobre un conjunto específico. Aunque alpha igual a 10 dé buenos resultados, no se puede confirmar que sea realmente óptimo sin explorar sistemáticamente otros valores como 8 o 15. Además, probar manualmente varios valores y elegir el mejor puede causar sobreajuste al conjunto de validación. GridSearchCV proporciona reproducibilidad científica, documentando automáticamente todos los valores probados y sus resultados. Finalmente, ofrece una estimación más confiable de cómo el modelo se comportará con datos completamente nuevos, garantizando una mejor generalización que la simple prueba manual de un único valor.

**7.** Además del modelo en sí (ej. Lasso()), ¿cuáles son los dos componentes principales que debes proporcionar a GridSearchCV para iniciar la búsqueda de hiperparámetros?

**RESPUESTA**

Los dos componentes principales que se deben proporcionar a GridSearchCV además del modelo son el diccionario de hiperparámetros y el número de folds para validación cruzada. El primero, llamado param_grid, es un diccionario que especifica qué hiperparámetros explorar y qué valores probar para cada uno, por ejemplo, alpha con valores 0.1, 1, 10 y 100. El segundo componente es cv, que indica en cuántas particiones se dividirán los datos para la validación cruzada, típicamente 5 o 10 folds. Estos dos elementos son esenciales porque param_grid define el espacio de búsqueda de hiperparámetros que se va a explorar, mientras que cv establece cómo se evaluará cada combinación de manera robusta mediante validación cruzada. Opcionalmente se puede agregar la métrica de evaluación con scoring, pero los dos componentes fundamentales son estos.

**8.** Si GridSearchCV selecciona un alpha muy pequeño (cercano a cero) como el mejor parámetro para tu modelo, ¿qué te sugiere esto sobre el nivel de sobreajuste que tenía tu modelo original sin regularizar?

**RESPUESTA**

Si GridSearchCV selecciona un alpha muy pequeño cercano a cero como el mejor parámetro, esto sugiere que el modelo original probablemente no tenía un problema significativo de sobreajuste. Un alpha cercano a cero implica que la penalización es mínima o casi inexistente, haciendo que el modelo regularizado se comporte muy similar a uno sin regularizar. Esto indica que el modelo ya generaliza bien por sí mismo y que no hay beneficio en penalizar los coeficientes. Probablemente se tiene suficientes datos en relación con la complejidad del modelo, las características son relevantes y hay poco ruido. Si se aumentara alpha, el rendimiento empeoraría, confirmando que el modelo necesita toda su capacidad actual. En conclusión, un alpha óptimo cercano a cero señala que el modelo original está bien calibrado y la regularización no es realmente necesaria para mejorar su desempeño.

**9.** En un árbol de decisión para optimizar la logística de un almacén, ¿qué podría representar un nodo hoja?

**RESPUESTA**

En un árbol de decisión para optimizar la logística de un almacén, un nodo hoja representa una decisión final o acción concreta a tomar. Por ejemplo, podría indicar la zona de almacenamiento asignada como "almacenar en zona A de alta rotación" o "almacenar en refrigeración", la prioridad de envío como "envío urgente" o "envío estándar", o el método de picking a utilizar. El nodo hoja es el resultado final al que se llega después de evaluar condiciones en los nodos internos como el peso del producto, su fragilidad o la urgencia del pedido, indicando específicamente qué acción tomar.

**10.** Un ingeniero crea un árbol para predecir fallos en una máquina. El árbol es extremadamente profundo y tiene reglas muy específicas como "Si la temperatura es 75.3°C y la vibración es 0.152 m/s² y el operador es Juan...". ¿Qué problema de ajuste es este y por qué no sería fiable en la práctica diaria de la planta?

**RESPUESTA**

Este es un problema claro de sobreajuste u overfitting. El árbol ha creado reglas excesivamente específicas que memorizan casos particulares del entrenamiento en lugar de capturar patrones generales. No sería fiable en la práctica diaria porque en la planta real las condiciones nunca serán exactamente iguales a las del entrenamiento. Por ejemplo, la temperatura raramente será exactamente 75.3 grados sino valores cercanos como 75.2 o 75.5, y el modelo no sabría cómo clasificarlos. Además, factores como el operador específico probablemente son coincidencias en los datos, no causas reales de fallos. El árbol debería aprender reglas más generales como temperatura mayor a 70 grados y vibración mayor a 0.15, que sean aplicables a múltiples situaciones reales y robustas ante variaciones normales en las condiciones operativas.

**11.** Al visualizar la "importancia de las características" de tu árbol, descubres que el "proveedor de materia prima" es la variable más importante. ¿Qué acción inmediata podrías tomar en la planta con esta información?

**RESPUESTA**

Al descubrir que el proveedor de materia prima es la variable más importante, se podrían tomar varias acciones inmediatas en la planta. Primero, auditar y evaluar los proveedores actuales para identificar cuáles están asociados con mejor calidad y cuáles generan problemas. Segundo, establecer controles de calidad más estrictos al recibir materia prima, especialmente de proveedores problemáticos. Tercero, considerar renegociar contratos o cambiar a proveedores más confiables si algunos están correlacionados con defectos o fallas. También se podría segregar los lotes por proveedor para mejor trazabilidad y ajustar parámetros de producción según el origen. Finalmente, desarrollar un programa de certificación de proveedores o comunicar directamente los hallazgos para trabajar colaborativamente en planes de mejora. **12.** Si tu árbol de decisión está clasificando perfectamente los datos históricos pero falla mucho con los datos de la última semana (sobreajuste), ¿qué parámetro de poda ajustarías primero para que generalice mejor?

**RESPUESTA**

Si el árbol está sobreajustado, el parámetro de poda que se debería ajustar primero es max_depth o profundidad máxima del árbol. Este parámetro es la opción más efectiva porque limita directamente cuántos niveles puede tener el árbol, evitando que cree reglas excesivamente específicas y profundas. Un árbol menos profundo es forzado a aprender patrones más generales en lugar de memorizar casos particulares de los datos históricos. Este control es intuitivo y fácil de interpretar, además de que un solo ajuste puede eliminar múltiples niveles de decisiones innecesarias. Otros parámetros alternativos incluyen min_samples_split que controla el número mínimo de muestras para dividir un nodo, o min_samples_leaf para las hojas. Sin embargo, max_depth es el punto de partida más fundamental y directo para combatir el sobreajuste en árboles de decisión.
